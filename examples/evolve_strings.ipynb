## Search Space and Genetic Algorithm Parameters

One of the interesting aspects of this experiment is how quickly the search space grows as the length of the target string increases.

If there are `N` possible characters and the target has `L` positions, the total number of possible strings is:

$$
N^L
$$

For example, with 27 possible characters and an 18-character target:

$$
27^{18} \approx 2.48 \times 10^{25}
$$

A brute-force algorithm would potentially have to search through an enormous number of possible strings.

The Genetic Algorithm does **not** enumerate this entire search space. Instead, it maintains a population of candidate solutions and uses fitness, selection, crossover, and mutation to gradually move the population toward the target.

### Population Size Matters

The number of individuals in each generation also affects the behavior of the algorithm.

For example:

- A population of `10` explores fewer candidate solutions per generation but is cheaper to evaluate.
- A population of `100` provides more diversity and more opportunities for useful genetic combinations.
- A population of `1000` explores even more candidates, but requires more computation per generation.

Therefore, there is a trade-off between **population size, genetic diversity, and computational cost**.

````python
import random
````

````python
target = "Survive the Fittest"
characters = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz "
````

````python
class DNA:
    def __init__(self,length):
        self.genes = [random.choice(characters) for _ in range(length)]
        self.fitness = 0
    def crossover(self, partner):
        child = DNA(len(self.genes))

        midpoint = random.randint(1, len(self.genes) - 1)

        for i in range(len(self.genes)):
            if i < midpoint:
                child.genes[i] = self.genes[i]
            else:
                child.genes[i] = partner.genes[i]

        return child
    def calculate_fitness(self, target):
        score = 0

        for i in range(len(target)):
            if self.genes[i] == target[i]:
                score += 1

        self.fitness = score / len(target)
    def mutate(self, mutation_rate):
        for i in range(len(self.genes)):
            if random.random() < mutation_rate:
                self.genes[i] = random.choice(characters)
        
````

````python
population = [DNA(len(target)) for _ in range(100)]
x = population[0]
x.calculate_fitness(target)
print(x.genes)
print(x.fitness)
````

````python
length = len(target)
for dna in population:
    score = 0

    for i in range(length):
        if dna.genes[i] == target[i]:
            score += 1

    dna.fitness = score / length

population[0].fitness
````

````python
mating_pool = []

for dna in population:
    n = int(dna.fitness * 100)

    for _ in range(n):
        mating_pool.append(dna)

parent1 = random.choice(mating_pool)
parent2 = random.choice(mating_pool)
while parent2 == parent1:
    parent2 = random.choice(mating_pool)
````

````python
child = parent1.crossover(parent2)
child.calculate_fitness(target)
child.fitness
````

````python
generation = 0

while True:
    generation += 1

    # Calculate fitness
    for dna in population:
        dna.calculate_fitness(target)

    # Build mating pool
    mating_pool = []

    for dna in population:
        n = int(dna.fitness * 100)

        for _ in range(n):
            mating_pool.append(dna)

    # Create next generation
    new_population = []

    for _ in range(100):
        parent1 = random.choice(mating_pool)
        parent2 = random.choice(mating_pool)

        child = parent1.crossover(parent2)
        child.mutate(0.01)

        new_population.append(child)

    population = new_population

    # Print every 10 generations
    if generation % 10 == 0:
        print("Generation:", generation)

    # Check if target found
    for dna in population:
        if "".join(dna.genes) == target:
            print("Target found!")
            print("Generation:", generation)
            print("Result:", "".join(dna.genes))
            break
    else:
        continue

    break
````
